```python
import requests
from dotenv import load_dotenv
import os
import pandas as pd
```

```python
load_dotenv()
```

```python
url = "https://v3.football.api-sports.io/fixtures?league=1&season=2022"
```

```python
api_key = os.getenv("API_KEY")
```

```python
response = requests.get(url, headers={"x-apisports-key": api_key})
data = response.json()
```

```python
data = response.json()
```

```python
data
```

```python
matches_2022 = {}
for match in data["response"]:

    item = {}
    match_id = match["fixture"]["id"]
    home_team = match["teams"]["home"]["name"]
    away_team = match["teams"]["away"]["name"]
    home_team_id = match["teams"]["home"]["id"]
    away_team_id = match["teams"]["away"]["id"]

    date = match["fixture"]["date"]
    referee = match["fixture"]["referee"]
    venue_name = match["fixture"]["venue"]["name"]
    venue_city = match["fixture"]["venue"]["city"]
    goals = match["goals"]

    item = {
        "home_team_name": home_team,
        "away_team_name": away_team,
        "home_team_id": home_team_id,
        "away_team_id": away_team_id,
        "home_team_goals": goals["home"],
        "away_team_goals": goals["away"],
        "date": date,
        "referee": referee,
        "venue_name": venue_name,
        "venue_city": venue_city,
    }

    matches_2022[match_id] = item
```

```python
matches_2022
```

```python
df = pd.DataFrame.from_dict(matches_2022, orient="index")
```

```python
df
```

```python
df.to_csv("matches_2022.csv", index=False)
```
